create numeric age column

```python
df["age_numeric"] = pd.to_numeric(df["age"], errors="coerce")
```

create an age category column that keeps the meaning of A and J

```python
def clean_age_category(value):
    value = str(value).strip().upper()

    if value == "A":
        return "ADULT"
    elif value == "J":
        return "JUVENILE"
    elif "A" in value and "J" in value:
        return "ADULT & JUVENILE"
    elif value.isdigit():
        return "NUMERIC AGE"
    else:
        return "UNKNOWN"

df["age_category"] = df["age"].apply(clean_age_category)
```

```python
df["age_category"].value_counts()
```

for the rows with actual ages, create proper age groups:

```python
bins = [0, 17, 24, 34, 44, 54, 64, np.inf]

labels = [
    "0-17",
    "18-24",
    "25-34",
    "35-44",
    "45-54",
    "55-64",
    "65+"
]

df["age_group"] = pd.cut(
    df["age_numeric"],
    bins=bins,
    labels=labels,
    include_lowest=True
)
```

## clean the date column

```python
df["date_raw"] = df["date"]
```

```python
def clean_date(row):
    value = str(row["date"]).strip()
    parts = value.split("/")

    try:
        # Example: 6/24
        if len(parts) == 2:
            month, day = map(int, parts)

            # First section of dataset = 2016
            # Second section = 2017
            year = 2016 if row.name < 379 else 2017

        # Example: 06/25/16 or 10/31/2017
        elif len(parts) == 3:
            month = int(parts[0])
            day = int(parts[1])
            year = int(parts[2])

            if year < 100:
                year += 2000

        else:
            return pd.NaT

        return pd.Timestamp(
            year=year,
            month=month,
            day=day
        )

    except:
        return pd.NaT
```

```python
df["date"] = df.apply(clean_date, axis=1)
```

## create time of day category

```python
def time_period(hour):
    if pd.isna(hour):
        return "UNKNOWN"
    elif 5 <= hour < 12:
        return "MORNING"
    elif 12 <= hour < 17:
        return "AFTERNOON"
    elif 17 <= hour < 21:
        return "EVENING"
    else:
        return "NIGHT"
```

```python
df["time_period"] = df["hour"].apply(time_period)
```

## clean cause code

```python
df["cause_code_raw"] = df["cause_code"]
```

convert to numeric

```python
df["cause_code"] = pd.to_numeric(
    df["cause_code"],
    errors="coerce"
).astype("Int64")
```

create broader groups

```python
def cause_group(code):

    if pd.isna(code):
        return "NOT RECORDED"

    elif 1 <= code <= 58:
        return "DRIVER / CYCLIST BEHAVIOUR"

    elif 59 <= code <= 68:
        return "PEDESTRIAN BEHAVIOUR"

    elif 69 <= code <= 73:
        return "PASSENGER / ATTENDANT"

    elif 74 <= code <= 75:
        return "ANIMAL"

    elif 76 <= code <= 77:
        return "OBSTRUCTION"

    elif 78 <= code <= 89:
        return "VEHICLE DEFECT"

    elif 90 <= code <= 93:
        return "ROAD CONDITION"

    elif 94 <= code <= 96:
        return "WEATHER"

    else:
        return "CAUSE NOT TRACED"
```

```python
df["cause_group"] = df["cause_code"].apply(cause_group)
```

```python
df["cause_group"].value_counts()
```
